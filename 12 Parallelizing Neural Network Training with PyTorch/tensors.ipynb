```python
import torch
import numpy as np

print('PyTorch version:', torch.__version__)

np.set_printoptions(precision=3)
a = [1, 2, 3]
b = np.array([4, 5, 6], dtype=np.int32)

t_a = torch.tensor(a)
t_b = torch.from_numpy(b)

print(t_a)
print(t_b)
```

```python
# Изменение типа данных тензора 
t_a_new = t_a.to(torch.int64)
print(t_a_new.dtype)
```

```python
# Транспонирование тензора
t = torch.rand(3, 5)
t_tr = torch.transpose(t, 0, 1)
print(t.shape, ' --> ', t_tr.shape)
```

```python
# Изменение формы тензора
t = torch.zeros(30)
print(t.shape)
t_reshape = t.reshape(5, 6)
print(t_reshape.shape)
```

```python
# Удаление ненужных измерений
t = torch.zeros(1, 2, 1, 4, 1)
t_sqz = torch.squeeze(t, 2)
print(t.shape, ' --> ', t_sqz.shape)
```

```python
torch.manual_seed(1)

# Создание тензора с равномерным распределением [-1; 1)
t1 = 2 * torch.rand(5, 2) - 1
# Создание тензора со стандартным нормальным распределением
t2 = torch.normal(mean=0, std=1, size=(5, 2))
```

```python
# Поэлементное произведение тензоров одинакового размера
t3 = torch.multiply(t1, t2)
print(t3)
```

```python
# Среднее - mean; сумма - sum; стандартное отклонение - std.
t4 = torch.mean(t1, axis=0)
print(t4)
```

```python
# Матричное произведение
t5 = torch.matmul(t1, torch.transpose(t2, 0, 1))
print(t5)
```

```python
# Вычисление Lp-нормы (здесь p=2) тензора
norm_t1 = torch.linalg.norm(t1, ord=2, dim=1)
print(norm_t1)
```

```python
# Split, stack, and concatenate tensors

# Разделение тензора на несколько
torch.manual_seed(1)
t = torch.rand(6)
print(t)

t_splits = torch.chunk(t, 3)
[item.numpy() for item in t_splits]
```

```python
# Разделение тензора на несколько с указанием размеров
torch.manual_seed(1)
t = torch.rand(5)
print(t)

t_splits = torch.split(t, split_size_or_sections=[3, 2])
[item.numpy() for item in t_splits]
```

```python
# Конкатенация тензоров
A = torch.ones(3)
B = torch.zeros(2)

C = torch.cat([A, B], axis=0)
print(C)
```

```python
# Стекирование тензоров
A = torch.ones(3)
B = torch.zeros(3)

S = torch.stack([A, B], axis=1)
print(S)
```
